#### Libs

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
custom_params = {"axes.spines.right": False, "axes.spines.top": False}
sns.set_theme(style="ticks", rc=custom_params)
```

#### Data

```python
df = (
    pd.read_csv('../data/demographics/info.csv', sep=';')
    .astype({'age': 'int'})
)
```

```python
df.shape
```

#### Descriptive analysis

##### Total of each group

```python
df.label.value_counts()
```

```python
df.label.value_counts(normalize=True).round(2)
```

##### Dataset

```python
distribution = pd.crosstab(
    df['dataset'],
    df['label'],
    normalize='columns'
)
```

```python
distribution.round(2)
```

```python
distribution.plot(
    kind="bar",
    figsize=(7, 3)
)
plt.title("Dataset distribution by label")
plt.xlabel("Dataset")
plt.ylabel("Proportion")
plt.legend(title="Label")
plt.xticks(rotation=0)
plt.show()
```

##### Age

```python
df.groupby('label').age.describe().round(2)
```

```python
plt.figure(figsize=(7, 3))
sns.boxplot(x='label', y='age', data=df)
plt.title('Age distribution by label')
plt.xlabel('Label')
plt.ylabel('Age')
plt.show()
```

```python
plt.figure(figsize=(7, 3))
sns.kdeplot(
    x='age',
    hue='label',
    fill=True,
    common_norm=False,
    alpha=0.5,
    data=df
)
plt.title('Age density by label')
plt.xlabel('Age')
plt.ylabel('Density')
plt.show()
```
